# Convert existing algorithm to an OGC Application Package

This page describes how to convert an existing algorithm into an OGC application package. There are two things the user will need to do:

- confirm their algorithm's run script is compliant
- re-register their algorithm

Prerequisite:
- an algorithm that is already deployed and has run successfully on MAAP

**1. Check your algorithm's run script**

> Your algorithm run script is the script that triggers execution of your algorithm. Convention on MAAP has been to name it something like *run.sh* and keep it at the root of your algorithm repo -- though neither of these are specifically necessary. **This script must now have named arguments.** For example, if you have a *run.sh* script that calls your python algorithm code like this:

> *my_algo.py INPUT1 INPUT2*

> You will need to update your *run.sh* script to have named arguments like this:

> *my_algo.py --my_input1 INPUT1 --my_input2 INPUT2*

> And then update your algorithm code to handle this accordingly.
