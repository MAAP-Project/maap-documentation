# OGC Software Overview

This page details what MAAP software versions are OGC-compliant and where they are available.

| Software Name | Version | OGC-compliant? | Installed where? |
|---------------|---------|----------------|------------------|
| maap-py | <= v4 | NO | ADE, MAAP Hub images (OPS & UAT) |
| maap-py | >= v5 | YES | MAAP Hub custom images (OPS & UAT) |
| MAAP API | <= v4 | NO | ADE, MAAP Hub (OPS & UAT) |
| MAAP API | >= v5 | YES | MAAP Hub (OPS & UAT) |
| MAAP Plugin: Algorithm Registration | v0| NO | ADE, MAAP Hub (OPS & UAT) |
| MAAP Plugin: Algorithm Registration | >= v1 | YES | MAAP Hub custom images (OPS & UAT) |
| MAAP Plugin: Jobs | <= v1 | NO | ADE, MAAP Hub (OPS & UAT) |
| MAAP Plugin: Jobs | >= v2 | YES | MAAP Hub custom images (OPS & UAT) |

For detailed instructions on selecting a workspace from our container registry, see [here](https://docs.maap-project.org/en/ogc/system_reference_guide/algorithm_registration.html#Container-URLs).

For a list of available workspaces, view our container registry [here](https://repo.maap-project.org/root/maap-workspaces/container_registry).
