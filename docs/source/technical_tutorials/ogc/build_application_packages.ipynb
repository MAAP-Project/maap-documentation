# Register Algorithm as an OGC Application Package

This tutorial describes how to build an OGC application package from an existing algorithm using the MAAP plugins. By the end of the tutorial, users will know how to:

- build an OGC application package
- view progress of the build
- view list of registered OGC application packages (algorithms)

Context:
OGC application packages are cross-platform, executable workflows. An OGC application package is composed of a CWL workflow and an executable (typically a Docker image). MAAP users can take their existing algorithms and convert them into OGC application packages to execute them across different platforms using a CWL workflow runner.

**Before you start:**

- You should have your algorithm code in a publicly-accessible repository (e.g. GitHub, GitLab).
- The command that invokes your algorithm must use named arguments. For MAAP users, this typically means making sure the algorithm command in your `run.sh` script resembles: 

`my_algo.py --my_input1 INPUT1 --my_input2 INPUT2` 

and not this: 

`my_algo.py INPUT1 INPUT2`


1. **Start a workspace on MAAP Hub**

> If you do not already have a workspace started, follow the instructions to do so [here](./start_hub_workspace.ipynb).

2. **Open the Algorithm Catalog MAAP plugin.**

    a. Click on the **Algorithm Catalog** card from the Launcher tab.

    <img src="../../_static/ogc/build_app_pack/algo_catalog_card.png" width="00"/>
    

    b. Click on the **Register New Algorithm** button.

    ![Register New Algorithm Button](../../_static/ogc/build_app_pack/register_algo_btn.png)


3. **Populate the form with algorithm information**

    Required fields are marked with a red asterisk. Contextual information is provided for all fields. Algorithm inputs are optional, but if provided, the input name, label, description, and type are all required fields for each input.

    ![Registration Form](../../_static/ogc/build_app_pack/algo_reg_form.png)

4. **Register algorithm**

    a. Click the **Register Algorithm** button at the bottom of the registration form.

    <img src="../../_static/ogc/build_app_pack/submit_reg_btn.png" width="300"/>

    If the registration request was successfully submitted, you will see two toasts at the bottom right corner of your workspace: one pointing to the build UI to monitor progress of the registration and the other informing you that a yaml file containing the information you provided was created. This file can later be used to re-register algorithms using the **Load Algorithm Configuration** feature at the top of the registration form.

    <img src="../../_static/ogc/build_app_pack/toasts.png" width="300"/>

    b. Click the **View build & deployment status** button to open the Build & Deployment UI to monitor the progress of your registration.

    <img src="../../_static/ogc/build_app_pack/build_deploy_ui.png" width="500"/>


5. **Confirm OGC application package was successfully generated**
    
Once the algorithm has been successfully built and deployed, navigate back to the **Algorithm Catalog** tab and confirm your algorithm is listed.
