# Start a Workspace on MAAP Hub

This page will show the user how to log into the MAAP Hub and start an OGC workspace.

Prerequisites: 
 - EarthDataLogin account
 - MAAP account

**1. Navigate to the MAAP Hub**

> Navigate to: https://hub.maap-project.org/

**2. Login**

> Click **Login with NASA EarthData** and enter your EDL credentials to login.
![NASA MAAP Login](../dps_tutorial/_static/ogc/nasa_maap_login.png)

**3. Select workspace environment & resources**

> Once logged in, you will be redirected to the Jupyter Hub server options page. This page allows you to configure a workspace to suit your needs. MAAP provides several pre-configured workspaces for your convenience. A list of these workspaces is shown in Table 1.

> *Table 1. Available MAAP workspace images*

> | Image Name | Image Path | Packages Installed |
> |------|------------|--------------------------|
> | pangeo | `` | [See environment.yaml](https://github.com/MAAP-Project/maap-workspaces/blob/develop/base_images/2i2c/pangeo/environment.yml) |
> | isce3 | `` | [See environment.yaml](https://github.com/MAAP-Project/maap-workspaces/blob/develop/base_images/2i2c/isce3/environment.yml) |
> | pytorch | `` | [See environment.yaml](https://github.com/MAAP-Project/maap-workspaces/blob/develop/base_images/2i2c/pytorch/environment.yml) |
> | tensorflow2 | `` | [See environment.yaml](https://github.com/MAAP-Project/maap-workspaces/blob/develop/base_images/2i2c/tensorflow2/environment.yml) |
> | r | `Not yet available` | '' |

> a. Select **Other** from the **Environment** dropdown.
> b. Copy an image path from Table 1 and enter it in the **Custom image** field.

![server options](../dps_tutorial/_static/ogc/workspace_config.png)


**4. Start the workspace**

> Click the **Start** button at the bottom of the page to start the workspace.

**Next steps**

The MAAP workspaces have the maap-py client library and MAAP plugins already installed. These allow users to interact with the MAAP backend to register algorithms and run jobs. Users who want the UI experience should use the MAAP plugins. Users who want a programmatic experience should use the maap-py client library.

For guidance on the OGC version of maap-py, see [here](../../system_reference_guide/maappy_algorithms_jobs.ipynb).

For guidance on how to use the algorithm registration UI, see [here](./build_application_packages.ipynb).

For guidance on how to use the jobs UI, see [here](./submit_job.ipynb).
