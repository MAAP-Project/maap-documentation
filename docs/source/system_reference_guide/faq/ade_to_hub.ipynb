# Migrating from ADE to Hub

In 2026, MAAP rolled a new code editing environment for developing algorithms. The new system referred to as the [MAAP Hub](hub.maap-project.org) is similar to the [pre-existing ADE](ade.maap-project.org), in that it's based around JupyterHub, however it has a few differences we want to highlight. This guide is intended for existing users who need help understanding the differences in the environments and how to transition to the MAAP Hub. It will not comprehensively cover all new features of the Hub as the rest of the documentation will be updated to cover general usage.

## Why are we changing the ADE to Hub?

- New features, a few highlights
    - Code Editors: VSCode, Rstudio, etc.. are launchable from within Workspaces
    - R environments are no longer `conda` based, should be much easier to install packages, Shiny apps work.
    - Custom Environments are self service so you can boot directly into custom images defined in github
    - GPU Support
    - Public Binder links to notebooks+environments: you can send a url to non-maap users with interactive code on a limited read-only instance.
    - QGIS Desktop image with web browser
- More efficient resource usage
    - Users can pick instance size at workspace launch, every time
    - We will turn off idle instance more frequently to save costs
    - User based limits on instance types 
- Ease of software maintenance
    - The new services use the same software as many other NASA -- VEDA, CryoCloud, GHG, Disasters, etc... -- and Non NASA hubs -- e.g. NOAA. This leads to collaboration and faster development of new features while working from a more standard JupyterHub base that is widely used.
    - Images are lightly customized from Pangeo Notebook and Rocker/py-rocket-geospatial community projects.

## Getting Started

**Users must transition to [https://hub.maap-project.org](https://hub.maap-project.org) before January 31, 2027. After January 31, 2027, you will no longer have access to [https://ade.maap-project.org](https://ade.maap-project.org).** Please reach out to the MAAP support team via Slack or email to activate your account.


### Tips

- Make sure to login with the same EarthDataLogin account you used in the ADE, otherwise you won't see files in your Buckets.
- Pick the smallest RAM/CPU allocation you need when booting a workspace. Smaller instance boot faster and cost less, so we can redirect that spend towards more compute.
- Want to learn how to create a custom image, ssh, or other advanced usage check out the [VEDA docs](https://docs.openveda.cloud/user-guide/scientific-computing/custom-environments.html)
- Have a general usage question?
    - 2i2c - https://docs.2i2c.org/
    - JupyterHub https://jupyterhub.readthedocs.io/en/stable/
    - Still not sure? Reach out in the MAAP Slack channels or via [Email](support@maap-project.org)



## Data Transition

Due to the new Quota system and existing directory sizes, we are not planning on moving Home Directory data across the platforms at this time, however your **buckets** will remain the same. At the time of writing this, MAAP has over 50 TB of files in home directories, but only 2 TB of that has been accessed in the last 60 days.

We recommend the following methods for moving your files off of the ADE:

### Code
 If you have any *code* (Jupyter notebooks, Python files, etc.), we always recommend that you keep it in version control (e.g. github, gitlab, etc...). You can always reclone repos in the new Hub as needed. For more information and examples about using Git with Github, see [About Git](https://docs.github.com/en/get-started/using-git/about-git).

### Data
If you have any *data* you need to keep, move it into `my-private-bucket` or `my-public-bucket`. Your buckets are identical across the platform, any files in these locations will be visible from the ADE and Hub. 

**Example code using Python:**

- Move individual files to your private bucket: `aws s3 mv path/to/file.csv s3://maap-ops-workspace/your-username/`

- Move folders to your private bucket: `aws s3 mv --recursive folder_name/ s3://maap-ops-workspace/your-username/folder_name/`

If you need to move folders into `my-public-bucket`, add `shared` between `maap-ops-workspace/` and `/your-username/`.

NOTE: `mv` moves your file into S3, and deletes it from the ADE. Replacing `mv` with `cp` will copy the data into S3, and not remove it from the ADE.




## Algorithms


Be aware that if you use DPS, there will be new implications to creating and registering algorithms in the MAAP Hub. Algorithms must be OGC-compliant in the MAAP Hub. For more information, see [OGC on the MAAP](https://docs.maap-project.org/en/ogc/technical_tutorials/ogc.html).

## Additional Hub Notes
We are aware that not everything will be the same and encourage you to ask questions and provide feedback. 

Examples of some parameters we plan to adjust based on user needs:

- Home Directory Quotas: the default is 150 GB, you still have access to Buckets for more storage.
    - At this time we are not moving your home directory data, see [Data Transition](ade_to_hub.ipynb#data-transition)
    - If you run out of space, we can grant temporary exceptions on a case by case basis. For more information on Hub storage, see [Disk Storage on MAAP](../disk_guide.ipynb)

- Workspace Timeouts: by default an idle workspace automatically turns off after and hour.

- Hot nodes: having already running machines to put workspaces on saves boot time but costs more, we need to tune this based on actual usage. See [Tips](#tips) about right-sizing your instance.
